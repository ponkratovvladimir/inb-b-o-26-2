## Ответы на контрольные вопросы

1)
Нумерация символов идёт с нуля, поэтому первый символ имеет индекс 0, а последний — на единицу меньше длины: 10 − 1 = 9.
2)
Python остановит программу с ошибкой - возникнет исключение IndexError с сообщением о том, что индекс вышел за пределы строки (string index out of range). В отличие от некоторых языков, символ «за краем» строка не вернёт - это именно ошибка выполнения.
3)
Встроенной функцией len(), передав ей строку: len('abcdefg') вернёт 7. Функция посчитает все символы строки, включая пробелы и знаки препинания.
4)
animal = 'Тигр'
animal(0) = 'Л'
Здесь сразу две ошибки:
animal(0) - это вызов функции с аргументом 0, а для доступа к символу нужны квадратные скобки: animal[0];
строки в Python неизменяемы: нельзя присвоить новое значение отдельному символу уже существующей строки. Запись animal[0] = 'Л' тоже вызвала бы ошибку TypeError.
Правильно так: собрать новую строку из частей - animal = 'Л' + animal[1:] - и она получится 'Лигр'.
5)
mystring = 'abcdefg'
print(mystring[2:5])
Выведет cde. Срез [2:5] берёт символы с индексами 2, 3 и 4 - правая граница не включается, а символы до неё идут без единицы.
6)
mystring = 'abcdefg'
print(mystring[3:])
Выведет defg. Если правую границу среза не указать, Python берёт всё до конца строки - то есть с индекса 3 (буква d) и дальше до последнего символа.
7)
mystring = 'abcdefg'
print(mystring[:3])
Выведет abc. Если в срезе не указана левая граница, Python берёт символы с самого начала строки, то есть с индекса 0, и идёт до правой границы не включительно - индексы 0, 1, 2.
8)
if 'd' in mystring:
    print("'d' входит в mystring")
else:
    print("'d' не входит в mystring")
Оператор in проверяет вхождение подстроки и возвращает True или False, поэтому его удобно использовать прямо в условии.
9)
little = big.lower()
Метод lower() возвращает копию строки, где все буквы переведены в нижний регистр, - и эту копию мы присваиваем переменной little. Сама строка big при этом не меняется, потому что строки в Python неизменяемы: любой строковый метод всегда возвращает новую строку.
