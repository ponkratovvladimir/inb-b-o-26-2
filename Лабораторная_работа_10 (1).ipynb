## Решение

Файл `pr7_00.txt` расположен репозитории. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

````python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/refs/heads/main/data/pract7/p7_data_00.csv
````

Прочитаем файл и превратим каждую строку в словарь (пользователя). Все пользователи хранятся в списке `peoples`:

````python
# список для хранения пользователей
peoples = []
with open('p7_data_00.csv', encoding='utf-8') as f:
    f.readline()  # пропускаем строку с заголовками
    for line in f:
        d = line.strip().split(',')
        user = {}
        # ID,Fname,Lname,Email,Gender,Salary,HasChildren
        user['ID'] = d[0]
        user['Fname'] = d[1]
        user['Lname'] = d[2]
        user['Email'] = d[3]
        user['Gender'] = d[4]
        user['Salary'] = float(d[5][1:])  # убираем знак валюты
        user['HasChildren'] = True if d[6] == 'true' else False
        peoples.append(user)
print(len(peoples))
````

Посмотрим на первый элемент списка:

````python
print(peoples[0])
````

### 1. Сортировка

В `sort` и `sorted` лямбда указывает **по какому полю** сравнивать элементы (параметр `key`).

**Задача 1.1.** Отсортировать пользователей по зарплате (по возрастанию) и вывести первых 10.

````python
# Сортировка по зарплате
peoples.sort(key=lambda u: u['Salary'])
for user in peoples[:10]:
    print(user['Fname'], user['Lname'], user['Salary'])
````

**Задача 1.2.** 10 пользователей с самой большой зарплатой (сортировка по убыванию).

````python
top = sorted(peoples, key=lambda u: u['Salary'], reverse=True)
for user in top[:10]:
    print(user['Fname'], user['Lname'], user['Salary'])
````

**Задача 1.3.** Сортировка по фамилии, а при одинаковых фамилиях — по имени (ключ — кортеж из двух полей).

````python
by_name = sorted(peoples, key=lambda u: (u['Lname'], u['Fname']))
for user in by_name[:10]:
    print(user['Lname'], user['Fname'])
````

### 2. Фильтрация

В `filter` лямбда возвращает `True` или `False`

**Задача 2.1.** Найдем всех мужчин без детей с зарплатой выше 5000.

````python
target = list(filter(lambda u: u['Gender'] == 'Male'
                     and not u['HasChildren']
                     and u['Salary'] > 5000,
                     peoples))
print('Найдено:', len(target))
for user in target[:10]:
    print(user['Fname'], user['Lname'], user['Gender'], user['Salary'], user['HasChildren'])
````

**Задача 2.2.** Найдем всех женщин, у которых есть дети.

````python
women_kids = list(filter(lambda u: u['Gender'] == 'Female' and u['HasChildren'], peoples))
print('Женщин с детьми:', len(women_kids))
````

**Задача 2.3.** Найдем пользователей, у которых электронная почта заканчивается на `.com`.

````python
com_users = list(filter(lambda u: u['Email'].endswith('.com'), peoples))
print('Почта на .com:', len(com_users))
````

### 3. Отображение

В `map` лямбда **превращает** каждый элемент в новый, и получается новый список.

**Задача 3.1.** Построить список, который содержит только ФИО и зарплату.

````python
fio_salary = list(map(lambda u: (u['Fname'] + ' ' + u['Lname'], u['Salary']), peoples))
print(fio_salary[:3])
````

**Задача 3.2.** Получить список электронных адресов заглавными буквами.

````python
emails = list(map(lambda u: u['Email'].upper(), peoples))
print(emails[:3])
````

**Задача 3.3.** Посчитать зарплату после вычета налога 13%.

````python
net = list(map(lambda u: (u['Lname'], round(u['Salary'] * 0.87, 2)), peoples))
print(net[:3])
````

### 4. Всё вместе

**Задача.** Найти 5 самых богатых женщин с детьми и вывести их в виде «Имя Фамилия: зарплата». Здесь используются сразу `filter`, `sorted` и `map`.

````python
rich_women = sorted(
    filter(lambda u: u['Gender'] == 'Female' and u['HasChildren'], peoples),
    key=lambda u: u['Salary'], reverse=True)[:5]

for line in map(lambda u: f"{u['Fname']} {u['Lname']}: {u['Salary']}", rich_women):
    print(line)
````

## Контрольные вопросы


**1. Что такое анонимная функция?**

Это функция без имени. Её пишут прямо в том месте, где она нужна, и обычно используют один раз. В Python анонимные функции создаются с помощью слова `lambda`.

**2. Опишите синтаксис объявления лямбда-выражения.**

`lambda аргументы: выражение`

Аргументы перечисляются через запятую, после двоеточия идёт одно выражение, значение которого функция возвращает (слово `return` писать не нужно). Например, `lambda x: x * 2` — то же самое, что `def f(x): return x * 2`.

**3. Поясните назначение лямбда-выражений. Приведите примеры их использования.**

Они нужны, чтобы коротко записать маленькую функцию и передать её в другую функцию, не придумывая ей имя и не создавая отдельный `def`. Примеры:
- сортировка: `sorted(words, key=lambda w: len(w))`
- фильтрация: `filter(lambda x: x > 0, nums)`
- отображение: `map(lambda x: x ** 2, nums)`

**4. Приведите пример использования лямбда-выражений для фильтрации элементов списка.**
nums = [1, 2, 3, 4, 5, 6]
even = list(filter(lambda x: x % 2 == 0, nums))
print(even)  # [2, 4, 6]
**5. Что будет напечатано в результате выполнения кода?**

```python
sx = '1.1 3.3 6.6'
x = list(map(float, sx.split()))
print(x[1])
```

`sx.split()` даёт список строк `['1.1', '3.3', '6.6']`, `map(float, ...)` превращает их в числа `[1.1, 3.3, 6.6]`, а `x[1]` — это второй элемент. **Ответ: `3.3`.** Проверим:
sx = '1.1 3.3 6.6'
x = list(map(float, sx.split()))
print(x[1])
**6. Что будет напечатано в результате выполнения кода?**

```python
data = [13, 1.3, 4, 'abs', 78]
filtered_data = list(filter(lambda x: type(x)==int, data))
print(filtered_data)
```

Фильтр оставляет только элементы типа `int`: `13`, `4` и `78` (`1.3` — это `float`, `'abs'` — строка). **Ответ: `[13, 4, 78]`.** Проверим:
data = [13, 1.3, 4, 'abs', 78]
filtered_data = list(filter(lambda x: type(x)==int, data))
print(filtered_data)
