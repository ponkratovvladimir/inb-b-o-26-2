# Ответы на вопросы
1. Чем отличается интерпретируемый язык от компилируемого? В компилируемом языке компилятор заранее переводит весь исходный код в машинный (или промежуточный) код, создавая исполняемый файл. Программа запускается уже из этого файла. В интерпретируемом языке нет отдельного этапа компиляции: специальная программа - интерпретатор - читает код построчно и сразу выполняет каждую инструкцию. Сейчас грань размывается: многие языки (например, Python, Java) используют гибридный подход (байт-код + JIT-компиляция), но классическое различие именно такое.

2. Компилируемые языки: C, C++, Rust, Go, Haskell, Erlang.

3. Интерпретируемые языки: Python, JavaScript, Ruby, PHP, Perl.

4. Как объявляются переменные в Python? В Python не нужно отдельно объявлять переменную - она создаётся в момент первого присваивания значения с помощью оператора =. Например: name = "Анна", age = 25.

5. Как определить тип переменной в Python? Используйте встроенную функцию type(). Она возвращает тип объекта, на который ссылается переменная. Например: print(type(x)) выведет <class 'int'>, если x = 5. Также есть функция isinstance() — она полезна, когда нужно проверить, принадлежит ли объект к классу или его подклассу (учитывает иерархию наследования).

6. Для чего используется функция print()? Для вывода информации на экран (в консоль). С её помощью показывают результаты работы программы, значения переменных, промежуточные данные для отладки. Можно передавать несколько аргументов, настраивать разделитель (sep), символ в конце строки (end), перенаправлять вывод в файл (file) и управлять буферизацией (flush).

7. Какая функция вычисляет модуль? Какую библиотеку подключать? Для вычисления модуля (абсолютного значения) числа используется встроенная функция abs(). Никакую библиотеку подключать не нужно - она доступна «из коробки». Например: abs(-3) вернёт 3. Модуль math нужен для других, более сложных математических операций (тригонометрия, логарифмы, корни).

8. Какие из перечисленных идентификаторов нельзя использовать? Нельзя использовать:

_1_01 - начинается с цифры (идентификатор должен начинаться с буквы или подчёркивания);
int - это ключевое слово языка (зарезервировано для своих целей);
create-var - содержит дефис (в именах разрешены только буквы, цифры и подчёркивание);
4perem - начинается с цифры;
System.Double - содержит точку (специальный символ, не разрешён).
Можно использовать: b100, _b200, MyVar, _5elem, zo0, wodoo, UserCount, system_call, string.

9. Основные типы в Python: int (целые числа), float (числа с плавающей точкой), str (строки), bool (логические значения: True, False), list (списки), tuple (кортежи), dict (словари), set (множества), None (специальное значение, обозначающее отсутствие значения).

10. Как называется программа, выполняющая специализированную задачу? Такая программа называется утилитой (utility). Вирусные сканеры, программы для сжатия файлов (архиваторы), инструменты резервного копирования - это тоже утилиты.
11. Прикладное программное обеспечение. Все перечисленные программы решают конкретные задачи пользователя: работа с текстом, таблицами, почтой, просмотр веб-страниц, развлечения.

12. 1 байт (8 бит). Этого достаточно, чтобы закодировать один символ из алфавита мощностью до 256 (например, латинские буквы, цифры, знаки препинания) или небольшое целое число.

13. Двоичная (бинарная) система счисления. В ней числа записываются последовательностями нулей и единиц.

14. Сопоставить каждому символу (букве, цифре, знаку препинания, управляющему символу) уникальный числовой код. Это позволяет компьютеру хранить и обрабатывать текст.

15. Юникод (Unicode). Этот стандарт включает знаки почти всех письменных языков мира (кириллица, иероглифы, арабская вязь и др.) и позволяет корректно отображать текст на разных языках.

16. Цифровые данные - информация, представленная в дискретной форме (в виде последовательности нулей и единиц). Цифровое устройство - устройство, которое принимает, хранит и преобразует такую дискретную информацию по определённому алгоритму.

17. Машинный код. Это набор инструкций (в виде двоичных кодов), который непосредственно выполняет центральный процессор.

18. Оперативная память (ОЗУ, RAM). При каждом запуске программы её исполняемый код и необходимые для работы данные загружаются из постоянной памяти (HDD/SSD) в оперативную, чтобы процессор мог быстро к ним обращаться.

19. Исполнение (выполнение) программы. В этом процессе ЦП последовательно извлекает инструкции из памяти, декодирует их (определяет, какую операцию нужно выполнить) и исполняет (проводит вычисления, перемещает данные и т.п.).

20. Язык ассемблера - это низкоуровневый язык программирования, где машинные инструкции (числа) заменены на мнемонические обозначения (короткие слова на английском, например MOV, ADD, SUB), понятные человеку. Он почти напрямую соответствует инструкциям конкретного процессора. Программа-ассемблер переводит такой код в машинный.
