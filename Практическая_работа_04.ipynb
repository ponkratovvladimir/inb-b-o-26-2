## Обработка символьной последовательности

Строки, подобно списка, являются коллекциями. Только в качестве элементов последовательности выступают символы. Для строк также используются функции нахождения длины `len()`, операция индексации `[]`

Например, можно посимвольно обходить строку, используя цикл `for` и индекс строки:

```python
# Вывод первых 10 символов строки с одновременным анализом буква это или цифра
for i in range(10):
  c = s[i]
  if c.isdigit():
    t = 'цифра'
  else:
    t='буква'
  print(c, t)
```

## Решение задачи 1

Расчитаем количество чисел кратных 3 но не кратным 9, и запишем их колличество, а также наибольшее из них

```python
import re

# Читаем содержимое файла
with open('data_v1.txt', 'r', encoding='utf-8') as f:
    content = f.read()

# Извлекаем все непрерывные последовательности цифр (числа)
numbers = re.findall(r'\d+', content)

# Преобразуем в целые числа
numbers_int = [int(n) for n in numbers]

# Фильтруем: кратные 3, но не кратные 9
filtered = [n for n in numbers_int if n % 3 == 0 and n % 9 != 0]

count = len(filtered)
max_num = max(filtered)

print(f"Количество чисел, кратных 3, но не кратных 9: {count}")
print(f"Максимальное из них: {max_num}")

```

# Отвты на вопросы
# 1)
Все три метода предназначены для чтения данных из открытого файлового объекта, но отличаются объёмом и форматом возвращаемых данных:
read(size=-1) - читает весь файл целиком и возвращает одну строку. Если указан аргумент size - читает не более size символов (байт). Удобен для небольших файлов, когда нужно получить всё содержимое разом.
readline(size=-1) - читает одну строку (до символа \n включительно). При повторном вызове продолжает со следующей строки. Если указан size - ограничивает длину считываемого фрагмента.
readlines() - читает весь файл и возвращает список строк, где каждый элемент — отдельная строка файла. По сути эквивалентно list(f).
Сходства: все возвращают данные из текущей позиции чтения, сдвигают позицию вперёд, возвращают пустую строку/список при достижении конца файла.
Различия: формат возврата (строка / одна строка / список строк) и объём данных, загружаемых в память за один вызов.
#2)
Файловый объект является итерируемым - по нему можно пройтись циклом for напрямую.Каждая итерация возвращает очередную строку. Также можно использовать list(f) или распаковку lines = [*f], но цикл for предпочтительнее - он не загружает весь файл в память разом.
#3)
Метод close() освобождает системные ресурсы, привязанные к файловому дескриптору, и сбрасывает буферизованные данные на диск. Если файл открыт для записи и не закрыт, часть данных может остаться в буфере и не записаться физически. Кроме того, операционная система имеет лимит на одновременно открытых дескрипторов - незакрытые файлы могут привести к ошибкам при превышении лимита.
Использование контекстного менеджера with гарантирует автоматическое закрытие файла даже при возникновении исключения:
#4)
Позиция считывания - это указатель (курсор) внутри файла, который определяет, с какого места будет читаться или записываться следующий байт/символ. Узнать её можно методом f.tell(), изменить - методом f.seek(offset).
При открытии файла в режиме чтения ('r') позиция изначально находится в начале файла (позиция 0). При открытии в режиме дозаписи ('a') - в конце файла.
#5)
Файл нужно открывать в режиме дозаписи — 'a' (от append).В этом режиме существующее содержимое сохраняется, а новые данные записываются в конец файла. Позиция записи всегда устанавливается в конец, даже если использовать seek() — при операции записи данные всё равно будут добавлены в конец (поведение зависит от платформы, но по стандарту так).
#6)
Исключение - это объект, который сигнализирует об ошибке или нештатной ситуации во время выполнения программы. Когда происходит ошибка (деление на ноль, обращение к несуществующему индексу, отсутствие файла и т. д.), Python создаёт объект исключения соответствующего типа. Если исключение не перехвачено, выполнение программы прерывается и выводится сообщение об ошибке с типом исключения и описанием.
#7)
Если исключение вызвано, а программа его не обрабатывает конструкцией try/except, происходит следующее:
1)Выполнение текущей функции прерывается.
2)Исключение распространяется вверх по стеку вызовов.
3)Если ни на одном уровне не находится обработчик - программа ** аварийно завершается**, и в консоль выводится traceback (трассировка стека) с типом исключения, сообщением и строкой кода, где оно произошло.
#8)
При попытке открыть несуществующий файл в режиме чтения Python вызывает исключение FileNotFoundError. Это подкласс OSError.
try:
    with open('nonexistent.txt', 'r') as f:
        content = f.read()
except FileNotFoundError:
    print("Файл не найден")

#10)
Если файла не существует, а программа пытается открыть его в режиме 'a' (дозапись), исключение не вызывается - файл автоматически создаётся.Если new_file.txt не было, он будет создан, и в него запишется "первая строка\n". Если файл уже существовал - данные добавятся в конец, а прежнее содержимое останется нетронутым.
То же самое верно и для режима 'w' (запись) - файл создаётся, если его нет. Разница в том, что 'w' обнуляет существующее содержимое, а 'a' - сохраняет.
