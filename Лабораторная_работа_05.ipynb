В файле pr5.txt содержится последовательность целых чисел.
Элементы последовательности могут принимать целые значения от

## Решение

Файл pr5.txt расположен репозитории. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

Реализуем чтение строк из файла и конвертацию отдельных строк в числовые значения. Каждое получаемое число будем добавлять в список `data`.

```python
# Читаем числа из файла pr5.txt
data = []
with open('pr5.txt', 'r', encoding='utf-8') as f:
    for line in f:
        line = line.strip()
        if line == '':  # пропускаем пустые строки
            continue
        data.append(int(line))

# Считаем среднее арифметическое всех чисел
avg = sum(data) / len(data)

count_pairs = 0
max_sum_pair = None  # будем хранить максимальную сумму подходящей пары

# Проходим по всем парам подряд идущих элементов
for i in range(len(data) - 1):
    a = data[i]
    b = data[i + 1]

    # Условие 1: хотя бы один из двух элементов меньше среднего
    cond_less_avg = (a < avg) or (b < avg)

    # Условие 2: хотя бы один из двух элементов оканчивается на 13
    # Для отрицательных чисел корректно проверяем последние две цифры
    def ends_with_13(x: int) -> bool:
        return abs(x) % 100 == 13

    cond_ends_13 = ends_with_13(a) or ends_with_13(b)

    if cond_less_avg and cond_ends_13:
        count_pairs += 1
        current_sum = a + b
        if max_sum_pair is None or current_sum > max_sum_pair:
            max_sum_pair = current_sum

# Если подходящих пар не нашлось, максимальную сумму можно вывести как 0 или None —
# в учебных задачах обычно подразумевается, что пара хотя бы одна есть.
if max_sum_pair is None:
    max_sum_pair = 0  # или можно оставить None, если в условии явно сказано, что пары есть

print(count_pairs, max_sum_pair)
```

Теперь ответим на поставленные вопросы использую список `data`

```python
# Вычислим среднее арифметическое всех чисел в файле
sr = sum(data) / len(data)
print(sr)
```

```python
counter = 0
max_sum = -20001

for i in range(len(data) - 1):
    x, y = data[i], data[i + 1]
    if (x < sr or y < sr) and (abs(x) % 100 == 13 or abs(y) % 100 == 13):
        counter += 1
        max_sum = max(max_sum, x + y)

print(counter, max_sum)
```

Но даже если написать names[0] = 'Вэнди' - это вызовет IndexError, потому что список пустой и индекс 0 ещё не существует.
Метод append() добавляет элемент в конец списка. Для пустого списка конец - это позиция 0, так что элемент займёт именно её. Альтернатива -names.insert(0, 'Вэнди'), но append() проще и естественнее, когда нужно просто добавить элемент в конец.

index(x) - возвращает индекс первого вхождения элемента x в списке. Если элемент не найден, вызывает ValueError.
insert(i, x) - вставляет элемент x в позицию с индексом i, сдвигая остальные элементы вправо. В отличие от append(), позволяет вставить элемент в любое место списка.
sort() - сортирует элементы списка по возрастанию (для чисел — по значению, для строк - по алфавиту). Изменяет сам список и возвращает None. Чтобы отсортировать по убыванию, нужно вызвать sort(reverse=True).
reverse() - меняет порядок элементов списка на обратный. Тоже изменяет сам список на месте и возвращает None. Не путать с sorted() и reversed(), которые создают новые объекты, не трогая исходный список.
