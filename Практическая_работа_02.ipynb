## Задание

Необходимо вычислить значение выражения:
$$Z=\frac{Y^2\cdot X}{2}-\frac{Y^2\cdot X^4}{4}+\frac{X^4\cdot Y^6}{6}-\frac{X^8\cdot Y^6}{8}+...$$
Для каждой переменной необходимо вывести ее значение и тип.

## Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: $X, Y$. Переменная, которая должна быть вычислена на основе независимых: $Z$.

Следует обратить внимание, что в выражении неизвестное число слагаемых, поэтому необходимо ввести переменную $N$ для представления количества слагаемых.

Обратите внимание, что знаки слагаемых чередуются, а степени числителя и множетели знаменателя привязаны к порядковому номеру слагаемого.

## Вычисление выражение, печать результатов

Код для вычисления значения выражения $Z$:

```python
X = 2
N = 10
Z = 0
for i in range(1, N+1):
  chisl = X**(i-1)
  znam = i * (i+1)
  if i%2==1:
    Z += chisl / znam
  else:
    Z -= chisl/znam
print(Z)
```

Решите пример в соответствии с индивидуальным заданием

```python
# Здесь код индивидуального задания...
X = 3
Y = 4
N = 12
Z = 0
for i in range(1, N+1):
  chisl = X**(i-1)
  chisl = Y**(i-1)
  znam = i * (i+1)
  if i%2==1:
    Z += chisl / znam
  else:
    Z -= chisl/znam
print(Z)
```

# Ответы на вопросы
1. Что такое переменная?
Переменная - это именованная область памяти, в которой хранится некоторое значение. В Python переменная не «содержит» данные напрямую, а ссылается на объект в памяти: имя переменной - это метка, по которой к объекту можно обратиться. Значение переменной можно менять в ходе работы программы.
2. Какие из приведённых имён переменных недопустимы в Python и почему?
99bottles - недопустимо: имя переменной не может начинаться с цифры.
july2009 - допустимо: начинается с буквы, дальше идут буквы и цифры.

3. Являются ли имена переменных Sales и sales одинаковыми? Почему?
Нет, они не одинаковы. Python - язык с учётом регистра (case-sensitive): Sales и sales - это две разные переменные. Интерпретатор воспринимает их как разные имена и будет хранить для них разные значения.
4. Допустима ли инструкция присваивания 72 = amount? Если нет, то почему?
Недопустима. В Python слева от оператора присваивания (=) должно стоять имя переменной, а справа - значение, которое ей присваивается. Здесь слева стоит число 72 (литерал), а ему нельзя «присвоить» что‑либо. Правильно было бы, например, amount = 72.

13. Что такое накопитель (аккумуляторная переменная)?
Накопитель (аккумулятор) - это переменная, которая накапливает промежуточный или итоговый результат в ходе выполнения цикла или последовательных операций. Чаще всего её используют для суммирования значений, подсчёта количества элементов или объединения строк. Например, если нужно сложить все числа из списка, в накопителе на каждом шаге будет храниться текущая сумма.
14. Следует ли инициализировать накопитель конкретным значением? Почему?
Да, обязательно. Накопитель нужно инициализировать перед началом работы (чаще всего нулём для суммы или пустой строкой для конкатенации). Причины:
Избежать ошибок. Если использовать переменную без предварительного присваивания, программа выдаст ошибку (например, NameError в Python).
Корректность вычислений. Без начального значения результат будет непредсказуемым. Например, при суммировании нужно стартовать с 0, иначе логика нарушится.
Ясность кода. Инициализация сразу показывает назначение переменной и её начальное состояние.

20. Что такое бесконечный цикл?
Бесконечный цикл - это цикл, который никогда не завершается сам по себе, потому что его условие продолжения всегда истинно. Такие циклы используют, когда нужно постоянно отслеживать события (например, ожидание ввода, опрос датчика, работу сервера), но в них обязательно предусматривают способ выхода (например, через break при определённом условии). Если выхода нет, программа будет работать, пока её не остановят принудительно (Ctrl+C и т. п.).
