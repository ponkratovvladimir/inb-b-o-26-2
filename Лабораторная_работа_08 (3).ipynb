## Задание

Определите количество чисел, пятеричная запись которых содержит ровно 5 цифр, причём каждая цифра отличается от соседних не менее, чем на 2.

## Решение

Ровно 5 цифр в пятеричной записи имеют числа от \(5^4 = 625\) до \(5^5 - 1 = 3124\): первая цифра не может быть нулём, иначе цифр в записи будет меньше пяти.

Алгоритм решения:
1. перебираем все числа от 625 до 3124;
2. каждое число переводим в пятеричную систему счисления;
3. проверяем условие: для каждой пары соседних цифр \(|a_i - a_{i+1}| \geq 2\);
4. если условие выполнено для всех пар — увеличиваем счётчик.

````python
def to_base5(n):
    """Перевод числа в пятеричную систему счисления (список цифр)."""
    digits = []
    while n > 0:
        digits.append(n % 5)
        n //= 5
    return digits[::-1]  # цифры от старшей к младшей

count = 0

for n in range(5**4, 5**5):   # от 625 до 3124 — все числа с ровно 5 цифрами
    digits = to_base5(n)
    ok = True
    for i in range(len(digits) - 1):
        if abs(digits[i] - digits[i+1]) < 2:
            ok = False
            break
    if ok:
        count += 1

print('Количество чисел:', count)
````

**Ответ: 140 чисел.**

Примеры подходящих чисел: 1010 → `13020`, 1014 → `13024`, 1015 → `13030`, 1020 → `13040`. Видна проверка соседних цифр, например, для 1010: |1−3| = 2, |3−0| = 3, |0−2| = 2, |2−0| = 2 — везде не меньше 2.

## Контрольная проверка другим способом

Проверим ответ методом динамического программирования: строим пятеричную запись цифра за цифрой и считаем количество строк фиксированной длины 5, где каждая следующая цифра отстоит от предыдущей не менее чем на 2. Первая цифра — от 1 до 4 (не ноль).

````python
from functools import lru_cache

@lru_cache(maxsize=None)
def f(pos, prev):
    if pos == 5:          # построили все 5 цифр
        return 1
    return sum(f(pos + 1, d) for d in range(5) if abs(d - prev) >= 2)

# первая цифра не ноль, иначе цифр было бы меньше 5
print('Контрольная проверка:', sum(f(1, first) for first in range(1, 5)))
````

Оба способа дают одинаковый результат — **140**, ответ надёжен.

# Контрольные вопросы

## 1. Что такое модуль в Python?

**Модуль** — это файл с расширением `.py`, содержащий исполняемый код на Python: функции, классы, переменные. Модуль служит единицей повторного использования кода и организации программы: вместо написания всего кода «с нуля» можно подключить готовый модуль. В Python модули могут быть:
- **встроенными** — входят в стандартную библиотеку (`math`, `random`, `itertools`, `os` и т. д.);
- **сторонними** — устанавливаются отдельно (`numpy`, `pandas`, `requests` и др.);
- **собственными** — создаются самим программистом как обычный `.py`-файл, который затем подключается по имени файла.

## 2. Сколько модулей может подключить программист?

Ограничения на количество подключаемых модулей нет — программист может подключить сколько угодно модулей. На практике подключают только те, которые нужны в конкретной программе, чтобы не расходовать память и не загромождать код.

## 3. С использованием какой команды производится подключение модуля?

Подключение производится командой **`import`**. Возможные варианты:

```python
import math                       # подключить модуль целиком
import math as m                  # подключить с псевдонимом
from math import sqrt             # импортировать конкретную функцию
from math import sqrt, pi         # несколько имён
from math import *                # все имена (не рекомендуется)
```

## 5. Перечислите несколько функций и классов модуля itertools, поясните их назначение.

- **`permutations(iterable, r)`** — возвращает итератор перестановок (все варианты упорядочивания) элементов; если `r` задан, то только перестановки длины `r`.
- **`combinations(iterable, r)`** — возвращает итератор сочетаний из `r` элементов без повторов (порядок не важен).
- **`combinations_with_replacement(iterable, r)`** — сочетания с возможностью повторения элементов.
- **`product(*iterables, repeat)`** — декартово произведение наборов; удобен для перебора всех комбинаций значений.
- **`chain(*iterables)`** — объединяет несколько последовательностей в одну без создания нового списка.
- **`accumulate(iterable)`** — накопительные суммы (кумулятивные итоги) элементов последовательности.
- **`count(start, step)`** — бесконечный счётчик чисел, начинающийся с `start` с шагом `step`.
- **`islice(iterable, start, stop)`** — «срез» итератора: берёт часть последовательности без копирования.
- **`groupby(iterable, key)`** — группирует последовательные элементы по значению ключа.
- **`cycle(iterable)`** — бесконечно повторяет элементы последовательности.
- **`repeat(obj, times)`** — повторяет объект указанное число раз.

Небольшая демонстрация:

````python
import itertools as it

print('permutations:', list(it.permutations('ABC', 2)))
print('combinations:', list(it.combinations('ABC', 2)))
print('product:     ', list(it.product([1, 2], 'ab')))
print('accumulate:  ', list(it.accumulate([1, 2, 3, 4])))
print('chain:       ', list(it.chain([1, 2], [3, 4])))
````

## 6. Укажите модули библиотеки Python, которые вы уже знаете. Опишите назначение и возможности данных модулей.

- **`math`** — математические функции: `sqrt` (квадратный корень), `sin`, `cos`, `log`, `factorial`, а также константы `math.pi` и `math.e`. Работает с действительными числами.
- **`random`** — случайные значения: `random.randint(a, b)` (случайное целое), `random.random()` (случайное число от 0 до 1), `random.choice(seq)` (случайный элемент), `random.shuffle(seq)` (перемешивание). Используется в играх, моделировании, генерации данных.
- **`itertools`** — использовался в этой работе: комбинаторные функции (`permutations`, `combinations`, `product`) и вспомогательные итераторы (`chain`, `accumulate`, `islice`). Позволяет красиво решать задачи перебора вариантов.
- **`os`** — работа с файловой системой: `os.path.exists(path)` (проверка существования файла), `os.listdir(dir)` (содержимое каталога), `os.mkdir` (создание каталога).
- **`sys`** — параметры запуска программы через `sys.argv`, управление стандартными потоками, выход из программы через `sys.exit`.
- **`datetime`** — получение текущей даты и времени (`datetime.now()`), разности дат, форматирование (`strftime`).
- **`csv`** — чтение csv-файлов объектами `csv.reader` и запись через `csv.writer`; удобнее ручного `split(',')`.
- **`collections`** — `Counter` (подсчёт частот), `defaultdict` (словарь со значением по умолчанию), `deque` (двусторонняя очередь).
- **`functools`** — декоратор `lru_cache` (кэширование вызовов функции, использовался выше для ДП), `partial` (частичное применение функции), `reduce` (свёртка последовательности).
- **`time`** — измерение времени работы кода (`time.time()`), паузы в выполнении программы (`time.sleep()`).
