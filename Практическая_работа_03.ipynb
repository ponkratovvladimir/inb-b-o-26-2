## Задание

Напишите функцию, которая по трем переданным им числам,
возвращает максимальное из них.

Напишем код для выичления переменных **a**, **b**, **с**

## Решение

### Объявление функции

```python
def sum_ab(a, b):
  s = 0
  for i in range(a, b+1):
    s+=i*i
  return s
```

### Демонстрация использования функции

```python
x, y = 1, 2
print(f'Сумма квадратов числел от {x} до {y}: {sum_ab(x, y)}')
x, y = 500, 50000
print(f'Сумма квадратов числел от {x} до {y}: {sum_ab(x, y)}')
x, y = 268, 3258
print(f'Сумма квадратов числел от {x} до {y}: {sum_ab(x, y)}')
```

```python
# Здесь код индивидуального задания...
def max_of_three(a, b, c):
    maximum = a
    if b > maximum:
        maximum = b
    if c > maximum:
        maximum = c
    return maximum
print(max_of_three(10, 3, 7))
print(max_of_three(-5, -2, -10))
print(max_of_three(4.5, 4.5, 3.2))
print(max_of_three(0, 0, 0))
```

С библиотечной функцией ты делаешь: «дай случайное число от 1 до 10», и получаешь число. А как именно генерируется случайность, какие алгоритмы и формулы используются - тебе не обязательно знать. Главное: вход → выход, предсказуемое поведение.

Аналогия: кнопка лифта. Ты нажимаешь этаж, лифт приезжает. Как он двигается, какие датчики и моторы работают - не важно для твоей задачи.
