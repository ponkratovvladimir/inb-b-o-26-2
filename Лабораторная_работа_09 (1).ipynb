## Задание

Разработать класс для представления объекта «Точка в пространстве». Реализовать:
* ввод и вывод полей данных;
* вычисление расстояния от точки до точки, заданной пользователем;

## Решение

Решение состоит из следующих этапов:
1. проектирование класса;
2. объявление класса (кодирование);
3. демонстрация использования класса (создание объекта и вызов методов).

На данном этапе спроектируем класс `Point3D`. У класса три закрытых поля данных (`__x`, `__y`, `__z`) - координаты точки, а также методы:
* `__init__()` - конструктор, вызывается автоматически при создании объекта;
* `SetPoint()` - метод-мутатор, записывает значения координат;
* `GetX()`, `GetY()`, `GetZ()` - методы-получатели, возвращают значения координат;
* `Input()` - ввод координат с клавиатуры;
* `DistToPoint()` - расстояние от точки до произвольной точки, заданной пользователем;
* `DistToOrigin()` - расстояние от начала координат;
* `__str__()` - возвращает информацию об объекте в виде строки.

Расстояние между двумя точками `A(x1, y1, z1)` и `B(x2, y2, z2)` вычисляется по формуле:

\[ d = \sqrt{(x_2-x_1)^2 + (y_2-y_1)^2 + (z_2-z_1)^2} \]

Расстояние от начала координат `O(0, 0, 0)` до точки `A(x, y, z)`:

\[ d_0 = \sqrt{x^2 + y^2 + z^2} \]

````python
import math

# Объявляем класс Point3D - точка в трехмерном пространстве
class Point3D:
  # Объявляем конструктор класса
  def __init__(self, px=0.0, py=0.0, pz=0.0):
    self.__x = px  # закрытое поле данных - координата x
    self.__y = py  # закрытое поле данных - координата y
    self.__z = pz  # закрытое поле данных - координата z

  # Метод-мутатор: записывает координаты в поля объекта
  def SetPoint(self, px, py, pz):
    self.__x = px
    self.__y = py
    self.__z = pz

  # Методы-получатели: возвращают значения координат
  def GetX(self):
    return self.__x

  def GetY(self):
    return self.__y

  def GetZ(self):
    return self.__z

  # Ввод полей данных с клавиатуры
  def Input(self):
    self.__x = float(input('Введите координату x: '))
    self.__y = float(input('Введите координату y: '))
    self.__z = float(input('Введите координату z: '))

  # Вывод полей данных на экран
  def Print(self):
    print('Координаты точки: (%.3f, %.3f, %.3f)' % (self.__x, self.__y, self.__z))

  # Расстояние от точки до точки, заданной пользователем
  def DistToPoint(self, px, py, pz):
    return math.sqrt((px - self.__x)**2 + (py - self.__y)**2 + (pz - self.__z)**2)

  # Расстояние от начала координат до точки
  def DistToOrigin(self):
    return math.sqrt(self.__x**2 + self.__y**2 + self.__z**2)

  # Возвращает информацию об объекте в виде строки
  def __str__(self):
    return ('Точка в пространстве:\n'
            '  Координаты: (%.3f, %.3f, %.3f)\n'
            '  Расстояние от начала координат: %.3f'
            % (self.__x, self.__y, self.__z, self.DistToOrigin()))
````

В данном листинге производится объявление типа данных `Point3D` с использованием зарезервированного слова `class`. Внутри класса объявляются функции и переменные. Поля `__x`, `__y`, `__z` объявлены с двумя нижними подчеркиваниями - это закрытые поля, доступ к ним извне класса ограничен.

Следует обратить внимание на использование зарезервированного слова `self`. Также внутри класса объявлен метод `__init__()`. Данный метод имеет специальное назначение — он вызывается в момент создания переменной данного класса автоматически. Данный метод называется конструктор класса.

# Демонстрация использования класса

Создадим объект `p1` типа `Point3D`, используя конструктор, и выведем информацию о нем.

Использование разработанного типа данных `Par`

````python
# Создание объекта p1 типа Point3D (используется конструктор)
p1 = Point3D(1.5, 2.0, -3.7)

# Вывод информации об объекте (автоматически вызывается метод __str__)
print(p1)
````

## Ввод полей данных и вычисление расстояний


````python
# Демонстрация ввода данных с клавиатуры
# (при запуске ячейки требуется ввести три числа)
p2 = Point3D()
p2.Input()
p2.Print()
print('Расстояние от начала координат: %.3f' % p2.DistToOrigin())
# Демонстрация без ввода с клавиатуры: используем метод-мутатор
p2 = Point3D()
p2.SetPoint(4.0, -1.0, 8.0)
p2.Print()

# Расстояние от точки p2 до точки, заданной пользователем (координаты 1, 2, 3)
d = p2.DistToPoint(1.0, 2.0, 3.0)
print('Расстояние до точки (1.000, 2.000, 3.000): %.3f' % d)

# Расстояние от начала координат
print('Расстояние от начала координат: %.3f' % p2.DistToOrigin())
````

## Проверка закрытости полей данных

Попытка обратиться к закрытому полю напрямую приведет к ошибке `AttributeError`, так как имя поля изменяется интерпретатором (механизм name mangling).

````python
# Попытка прямого доступа к закрытому полю
try:
    print(p1.__x)
except AttributeError as e:
    print('Ошибка доступа:', e)

# Корректный доступ - только через метод-получатель
print('Координата x точки p1:', p1.GetX())
````

Следует обратит внимание на объявление метода `Load()` в классе Par. Несмотря на его принадлежность классу, сам метод не содержит параметра `self`.

Это значит, что метод не принадлежит конкретному экземпляру (объекту), он принадлежит самому классу. Такие методы называют статическими. Для вызова статического метода нет необходимости явно создавать объект класса, используя конструктор. Статический метод вызывается из класса.

# Контрольные вопросы



**1. Дайте определение понятию класс.**

Класс - это описание структуры и поведения объектов: тип данных, который объединяет в себе данные (поля, атрибуты) и методы их обработки. Класс выступает шаблоном (чертежом), по которому создаются конкретные объекты — экземпляры класса.

**2. Использование метода `__init__` и его назначение.**

Метод `__init__` - конструктор класса. Он вызывается автоматически в момент создания объекта и служит для инициализации полей данных начальными значениями. Первый параметр метода — `self` - ссылка на создаваемый объект.

```python
class Point3D:
  def __init__(self, px=0.0, py=0.0, pz=0.0):
    self.__x = px
    self.__y = py
    self.__z = pz

# конструктор вызывается автоматически при создании объекта
p = Point3D(1.0, 2.0, 3.0)
```

**3. Назначение статических методов и их отличия от методов класса.**

Статический метод (объявляется с декоратором `@staticmethod`) не получает ни `self`, ни `cls` - он принадлежит классу, а не конкретному объекту, и не имеет доступа к полям экземпляра. Используется для вспомогательных операций, логически связанных с классом. Метод класса (`@classmethod`) в качестве первого параметра получает `cls` - сам класс, и имеет доступ к атрибутам класса. Обычный метод экземпляра получает `self` и работает с полями конкретного объекта.

```python
class Demo:
  @staticmethod
  def stat():      # не получает ни self, ни cls
    return 'статический метод'

  @classmethod
  def cls_m(cls):  # получает класс cls
    return 'метод класса'
```

**4. Для чего используется параметр `self` в каждом методе класса?**

Параметр `self` - ссылка на конкретный экземпляр класса, для которого вызван метод. Через `self` метод обращается к полям и другим методам этого объекта (`self.__x`, `self.GetX()`). При вызове `p1.Info()` интерпретатор автоматически передает в метод ссылку на объект `p1`. Без `self` метод не смог бы отличить поля одного объекта от полей другого.

**5. Сколько методов `__init__` может быть объявлено внутри класса?**

Формально объявить можно сколько угодно, но использоваться будет только последний объявленный: каждое следующее объявление `__init__` перезаписывает предыдущее (в Python отсутствует перегрузка методов). Поэтому фактически в классе работает один конструктор. Разные варианты инициализации реализуют через значения по умолчанию аргументов.

**6. Каким образом в классе Python атрибут скрывается от внешнего кода?**

Если имя атрибута начинается с двух подчеркиваний (`__x`), интерпретатор применяет механизм name mangling: имя внутри класса преобразуется в `_ИмяКласса__x`, и прямой доступ `obj.__x` извне класса вызывает ошибку `AttributeError`. Доступ организуют через методы-получатели и методы-мутаторы. Имя с одним подчеркиванием (`_x`) — лишь соглашение «не трогать снаружи», реальной защиты оно не дает.

```python
class Point3D:
  def __init__(self, px):
    self.__x = px      # закрытое поле

p = Point3D(1.0)
# p.__x              -> AttributeError
p._Point3D__x        # так поле все же доступно: 1.0
```

**7. Какова задача метода `__str__()`?**

Метод `__str__` возвращает строку — удобное для человека текстовое представление объекта. Он определяет, как объект будет выглядеть при выводе на экран функцией `print()` и преобразовании `str()`.

**8. Каким образом происходит вызов метода `__str__()`?**

Явно его вызывать не нужно: метод вызывается автоматически при передаче объекта функциям `print()` и `str()`, а также в f-строках и форматировании. Допустим и прямой вызов `p1.__str__()`, но так писать не принято.

```python
p1 = Point3D(1.5, 2.0, -3.7)
print(p1)            # __str__ вызывается автоматически
s = str(p1)          # и здесь тоже
```

**9. Что такое атрибут экземпляра?**

Атрибут экземпляра — поле данных (или метод), принадлежащее конкретному объекту, а не классу в целом. Такие атрибуты обычно создаются в конструкторе через `self.имя = значение`, и у каждого экземпляра они свои: у точки `p1` своя координата `x`, у точки `p2` — своя.

**10. Что такое метод-получатель? Что такое метод-мутатор?**

Метод-получатель (getter) — метод, возвращающий значение поля данных объекта, не изменяя его (например, `GetX()`). Метод-мутатор (setter) — метод, изменяющий значение поля данных (например, `SetPoint()`). Такая пара методов реализует инкапсуляцию: поля класса закрыты, а чтение и изменение значений происходит через контролируемый интерфейс.
